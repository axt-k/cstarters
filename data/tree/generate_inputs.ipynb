```python
df = load_dataset(path = os.path.join(data_dir, "cstarters_dataset.csv"))
with open('tree_alignment.fasta', 'w') as fasta_file:
    for k, v in outgroup_alignments.items():
        fasta_file.write(f">{k}\n{v}\n")
    for i, data in df.iterrows():
        fasta_file.write(f">{data['cs_accession']}\n{data['aligned_sequence']}\n")
with open('tree_data_chiplot.csv', 'w') as chiplot_file:
    chiplot_file.write('Node IDs,Category,3OH\n')
    for i, data in df.iterrows():
        category = data['starter_category']
        acc = data['cs_accession']
        hydroxylation = ''
        if data['starter_beta_hydroxylation'] == "TRUE":
            hydroxylation = 'True'
        chiplot_file.write(f"{acc},{category},{hydroxylation}\n")
```

```python
with open('tree_data.csv', 'w') as treedata_file:
    treedata_file.write('accession,category,3OH\n')
    for i, data in df.iterrows():
        category = data['starter_category']
        acc = data['cs_accession']
        hydroxylation = ''
        if data['starter_beta_hydroxylation'] == "TRUE":
            hydroxylation = '1'
        if data['starter_beta_hydroxylation'] == "FALSE":
            hydroxylation = '0'
        treedata_file.write(f"{acc},{category},{hydroxylation}\n")
```
