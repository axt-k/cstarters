```python
import os

import pandas as pd
```

```python
data_dir = "../../data"
df = pd.read_csv(os.path.join(data_dir, 'cstarters_dataset.csv'))

cs_acc_to_categories = {}
for i in range(len(df)):
    if df.iloc[i]['starter_category'] in ['SCFA', 'MCFA', 'LCFA', 'AR', 'FA']:
        if not df.iloc[i]['cs_accession'] in  cs_acc_to_categories:
            cs_acc_to_categories[df.iloc[i]['cs_accession']] = []
        if df.iloc[i]['starter_category'] not in cs_acc_to_categories[df.iloc[i]['cs_accession']]:
            cs_acc_to_categories[df.iloc[i]['cs_accession']].append(df.iloc[i]['starter_category'])

mibig_acc_to_categories = {}
for i in range(len(df)):
    if df.iloc[i]['starter_category'] in ['SCFA', 'MCFA', 'LCFA', 'AR', 'FA']:
        if not df.iloc[i]['mibig_v4_accession'] in  mibig_acc_to_categories:
            mibig_acc_to_categories[df.iloc[i]['mibig_v4_accession']] = []
        if df.iloc[i]['starter_category'] not in mibig_acc_to_categories[df.iloc[i]['mibig_v4_accession']]:
            mibig_acc_to_categories[df.iloc[i]['mibig_v4_accession']].append(df.iloc[i]['starter_category'])

compounds_num = 0
for i in range(len(df)):
    if df.iloc[i]['starter_category'] in ['SCFA', 'MCFA', 'LCFA', 'AR', 'FA']:
        compounds_num += 1

unique_cs_accs = list(cs_acc_to_categories.keys())
unique_mibig_accs = list(mibig_acc_to_categories.keys())

print(f"Number of unique CS accessions: {len(unique_cs_accs)}")
print(f"Number of compounds: {compounds_num}")
print(f"Number of unique MIBiG accessions: {len(unique_mibig_accs)}")
```
